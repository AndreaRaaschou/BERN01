### 2.3 Data cleaning

**My own notes:**
Looking at the data, I will try to merge the predictors (content of soil_df and stats dictionary) with the observed biome (from lpj_df). 

The soil_df has some problems. There is one row with values missing for lat and all soil metrics. I is probably easiest to completely discard this row when merging all dfs. soil_df and ljp_df are not lined up in the same order. Use location and not index to line merge the dfs. stats and lpj_df line up perfectly.

```python
# Compare shape of dfs
print(lpj_df.shape)
print(soil_df.shape)
print(stats_df["pre"].shape)
```

```python
# Check if lat and lon locations match
print((lpj_df["Lon"] == soil_df["lon"]).all())
print((lpj_df["Lat"] == soil_df["lat"]).all())

print((lpj_df["Lon"] == stats_df["pre"]["Lon"]).all())
print((lpj_df["Lat"] == stats_df["pre"]["Lat"]).all())
```

```python
# Check for values that are not a number
print(soil_df[["lon", "lat"]].isna().sum())
print(lpj_df[["Lon", "Lat"]].isna().sum())
```

```python
# Fix problems with soil_df
soil_df = soil_df.dropna() # drops all rows with nan values
soil_df[soil_df["lon"].isna()]
soil_df["lon"] = pd.to_numeric(soil_df["lon"]) # Change the lon column values from type str to float
soil_df.shape # check how many rows were dropped
soil_df = soil_df.rename(columns={"lon": "Lon","lat": "Lat"}) # Rename columns
```

### 2.4 Prepare one collected df
In this section, I will put the needed data from soil_df, stats and lpj_df into one big df that I can use in downstream analysis.

```python
# Start by taking wanted columns from lpj_df
collected_df = lpj_df[["Lon", "Lat", "Biome_obs"]].copy()

# Merge with soil_df
collected_df = collected_df.merge(
    soil_df,
    on=["Lon", "Lat"],
    how="left"
)

# Merge with stats (climate) as well
for name, df in stats_df.items():
    collected_df = collected_df.merge(
        df,
        on=["Lon", "Lat"],
        how="left",
        suffixes=("", f"_{name}") # add suffixes to keep climate columns apart
    )

collected_df.head()
```
